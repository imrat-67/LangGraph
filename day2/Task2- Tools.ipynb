```python
from langchain_core.tools import StructuredTool
from pydantic import BaseModel, Field
```

```python
class MultiplyInput(BaseModel):
    a: int = Field(required = True, description="The first number to add")
    b: int = Field(required = True, description="The seond number to add")
```

```python
def multiply_func(a:int , b:int) ->int:
    return a*b
```

```python
mutliply_tool = StructuredTool.from_function(
    func = multiply_func,
    name = "multiply",
    description="Multiply two numbers",
    args_schema=MultiplyInput
)
```

```python
result = mutliply_tool.invoke({'a':3, 'b':3})

print(result)
print(mutliply_tool.name)
print(mutliply_tool.args)
```

```python
from langchain.tools import BaseTool
from typing import Type
```

```python
class MultiplyInput(BaseModel):
    a: int = Field(required=True, description="The first number to add")
    b: int = Field(required=True, description="The second number to add")
```

```python
class MultiplyTool(BaseTool):
    name: str = "multiply"
    description: str = "Multiply two numbers"

    args_schema: Type[BaseModel] = MultiplyInput

    def _run(sef, a:int, b:int) -> int:
        return a * b
```

```python
mutliply_tool = MultiplyTool()
```

```python
result = mutliply_tool.invoke({'a':3, 'b':3})

print(result)
print(mutliply_tool.name)
print(mutliply_tool.description)
print(mutliply_tool.args)
```

```python
from langchain_core.tools import tool

@tool
def add(a:int, b:int) -> int:
    """Add two numbers"""
    return a + b

@tool
def multiply(a:int, b: int) -> int:
    """Multiply two numbers"""
    return a * b
```

```python
class MathToolkit:
    def get_tools(self):
        return [add, multiply]
```

```python
toolkit = MathToolkit()
tools = toolkit.get_tools()

for tool in tools:
    print(tool.name, "=>", tool.description)
```

```python
add_tool = toolkit.get_tools()[0]
print(add_tool.name, "=>", add_tool.description)
result = add_tool.invoke({"a":4, "b":9})

print(result)
```
